# Interaction gene symbols - September 25, 2015
Write gene symbols pairs for all interactions based on the UniProt entry name
## Configuration

```python
import pandas as pd
import os

data_dir = os.path.join('..','..','synthesis072614')
```

## Load all files

```python
uniprot_df = pd.read_csv(os.path.join(data_dir,'cytoscapeAnnotationsUniprot020815.tab'), sep='\t')
edges_df = pd.read_csv(os.path.join(data_dir,'p-0.01_202dir_cytoscape.txt'), sep='\t', header=None)

print '%d UniProt ids loaded' % len(uniprot_df)
print '%d interactions loaded' % len(edges_df)

# Create dictionaries of peptide to modified site for each replicate
uniprot_dict = dict(zip(uniprot_df['Entry name'], uniprot_df['Gene names  (primary )']))
```

## Lookup and add gene symbols, write the updated file
Do not write column headers or row indices

```python
edges_df['Gene1'] = edges_df.apply(lambda row: uniprot_dict[row[0]], axis=1)
edges_df['Gene2'] = edges_df.apply(lambda row: uniprot_dict[row[1]], axis=1)

edges_df.to_csv(os.path.join(data_dir,'p-0.01_202dir_cytoscape_genes_092515.txt'), sep='\t', header=False, index=False)
```
