```python
%pylab inline
import pyupset as pyu
import pandas as pd
```

```python
df_a = pd.DataFrame(data=pd.Series(data=['A','B','C']),columns=['Items'])
df_b = pd.DataFrame(data=pd.Series(data=['B','C','D','E']),columns=['Items'])
df_c = pd.DataFrame(data=pd.Series(data=['E','F','G']),columns=['Items'])
```

```python
pyu.plot({'Set A':df_a, 'Set B':df_b, 'Set C':df_c});
```
