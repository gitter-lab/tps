# PPI network statistics - August 20, 2015
Compute the number of edges, number of unique proteins, and number of protein prizes that are in the network

```python
import os

base_dir = os.path.join('..','..')
ppi_file = os.path.join(base_dir, 'interactomes', 'PhosphoSite_iRefIndex13.0_uniprot_overwrite_pcsf.txt')
prize_file = os.path.join(base_dir, 'phosphorylation-data-sets', 'McKenna2014', 'McKenna2014_aggregated_replicateABC_tukey_firstprev_proteinPrizes_firstProtein.txt')
```

## Load the PPI network and the protein prizes
Then print the interaction of the prize and PPI network proteins

```python
# A set with the names of all proteins that have prizes
prize_prots = set()
with open(prize_file) as pf:
    for line in pf:
        parts = line.upper().strip().split("\t")
        if len(parts) != 2:
            raise RuntimeError("All prize file lines must have 2 columns\n%s" % line)
        prize_prots.add(parts[0])

# A set with the names of all proteins in the PPI network
ppi_prots = set()
# Count the directed and undirected interactions in the PPI network
undir_count = 0
dir_count = 0

with open(ppi_file) as pf:
    for line in pf:
        parts = line.upper().strip().split("\t")
        if len(parts) != 4:
            raise RuntimeError("All prize file lines must have 4 columns\n%s" % line)
        ppi_prots.add(parts[0])
        ppi_prots.add(parts[1])
        
        if parts[3] == "D":
            dir_count += 1
        elif parts[3] == "U":
            undir_count += 1
        else:
            raise RuntimeError("Invalid interaction type\n%s" % line)

print "%d proteins in the prize file" % len(prize_prots)
print "%d proteins in the PPI network" % len(ppi_prots)
print "%d proteins in the prize file and the PPI network\n" % len(prize_prots.intersection(ppi_prots))

print "%d undirected interactions (may include self-edges)" % undir_count
print "%d directed interactions (may include self-edges)" % dir_count
print "%d total interactions (may include self-edges)" % (undir_count + dir_count)
```
