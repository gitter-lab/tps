# Permute the protein-peptide mapping - November 11, 2016
Take the TPS peptide-protein map file and create multiple permuted copies

```python
import random as rn
import os

rn.seed(111116)

in_dir = os.path.join('..','tps','data','timeseries')
map_file = os.path.join(in_dir, 'peptide-mapping.tsv')
```

## Test the shuffle method
This will give the same permuations after restarting the kernel due to the seed

```python
orig = range(5)
print 'Original list: {}'.format(orig)

for perm in range(10):
    copy = orig
    rn.shuffle(copy)
    print 'Permuatation {}: {}'.format(perm, copy)
```

## Load the protein-peptide map
Do not create a dictionary because we want to shuffle the values.  Do not split the set of proteins on a single line, they will be shuffled together.

```python
peptides = []
proteins = []

# Load two lists
with open(map_file) as map_f:
    # Skip the header
    map_f.readline()
    for line in map_f:
        parts = line.strip().split("\t")
        assert len(parts) == 2, "Expected tab-delimited peptide and protein (or pipe-delimited proteins) on each line"

        peptides.append(parts[0])
        proteins.append(parts[1])

for i in range(5):
    print '{}\t{}'.format(peptides[i], proteins[i])
```

## Shuffle the protein list

```python
test_peptide = 'K.n[305.21]EIPAS[167.00]PSR.T'

for shuffle in range(5):
    print 'Shuffle {}'.format(shuffle)
    proteins_copy = proteins
    rn.shuffle(proteins_copy)
    pep_map = dict(zip(peptides, proteins_copy))
    
    for i in range(5):
        print '{}\t{}'.format(peptides[i], proteins_copy[i])
    print pep_map[test_peptide]
```
