# Compute a correlation network for the synthesis toy example data

# Initialize

```python
import numpy as np
import os
toyDir = os.path.join('..','..','toyExamples','synthesis')
tsFile = os.path.join(toyDir,'TimeSeries.txt')
outSif = os.path.join(toyDir, 'toy.sif')
outAnnotation = os.path.join(toyDir, 'toyEdgeAnnotation.txt')
```

## Test numpy correlation with lists

```python
l1 = [0.9, 0.8, 0.7, 0.6]
l2 = [0.89, 0.77, 0.8, 0.58]
np.corrcoef(l1,l2)
```

## Compute correlation for all pairs of nodes

Load the array twice, once to get the node names and once to get the data.  There are better ways to do this but this is simple.  Then write all pairwise correlations (magnitude only) and a sif file with the edges.

```python
tsNodes = np.loadtxt(tsFile, delimiter='\t', skiprows=1, usecols=[0], dtype='string')
tsData = np.loadtxt(tsFile, delimiter='\t', skiprows=1, usecols=[1,2,3,4])
tsCorr = np.corrcoef(tsData)

with open(outSif, 'w') as sifWriter:
    with open(outAnnotation, 'w') as annWriter:
        annWriter.write('Interaction\tAbsoluteCorrelation\n')
        for i in range(len(tsNodes)):
            for j in range(i+1,len(tsNodes)):
                sifWriter.write('%s\tNone\t%s\n' % (tsNodes[i], tsNodes[j]))
                annWriter.write('%s (None) %s\t%f\n' % (tsNodes[i], tsNodes[j], abs(tsCorr[i, j])))
```
