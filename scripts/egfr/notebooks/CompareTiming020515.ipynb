# Compare times of significant peptide changes vs times of active (activation/inhibition/ambiguous) protein state - February 5, 2015

## Initializations

The peptide timing is derived from the Tukey HSD test, taking the most significant change with respect to 0 min or the previous time point.  Protein timing is the activity state for proteins in the Refine pathway.

```python
%cd ../python

import sys, os, csv
import EGFRUtil

baseDir = os.path.join('..','..')
workingDir = os.path.join(baseDir, 'synthesis072614')
pepTimingFile = os.path.join(baseDir, 'phosphorylation-data-sets', 'McKenna2014', 'aggregated_replicateABC_tukey_orig_firstprev.txt')
protTimingFile = os.path.join(workingDir, 'cytoscapeAnnotations020715.txt')
timingTableFile = os.path.join(workingDir, 'comparePepProtTiming020515.txt')
```

## Load the temporal peptide significance and temporal protein activity states 

```python
with open(pepTimingFile) as pepTimingF:
    reader = csv.DictReader(pepTimingF, delimiter = '\t')
    pepTiming = list(reader)

with open(protTimingFile) as protTimingF:
    reader = csv.DictReader(protTimingF, delimiter = '\t')
    protTiming = list(reader)
```

## Count the total peptides, significant peptides, and significant peptides at each time point

The significant peptides at each time point are the number of that are <= the 0.01 p-value threhsold.  Also track the number of peptides that are below that threshold at any time point.

```python
totPeptides = len(pepTiming)
sigPeptides = sum([1 for row in pepTiming if float(row['min.tukey.pval']) <= 0.01])
sigPeptidesTiming = dict()
for i in range(7):
    time = 2**(i+1)
    col = 'tp%d' % (i+2)
    sigPeptidesTiming[time] = sum([1 for row in pepTiming if float(row[col]) <= 0.01])
print 'Total peptides: %d\nSignificant peptides: %d\nTemporal significance: %s' % (totPeptides, sigPeptides, str(sigPeptidesTiming))
```

## Count the number of peptides that are first significant at each time point

```python
sigPeptidesFirstList = list()
for row in pepTiming:
    pvals = [float(row['tp%d' % (i+2)]) for i in range(7)]
    sigPeptidesFirstList.append(EGFRUtil.FirstSignificant(pvals, 0.01))
    
sigPeptidesFirst = dict()
for i in range(7):
    time = 2**(i+1)
    sigPeptidesFirst[time] = sum([1 for firstTime in sigPeptidesFirstList if firstTime == str(time)])
print '%d peptides do not have a time point at which they are first significant' % sum([1 for firstTime in sigPeptidesFirstList if firstTime == "Not significant"])
```

## Count the number of proteins in a category and the number of proteins in that category that are not inactive (activation/inhibition/ambiguous) at each time point.

The categories are all those in the Refiner pathway, all that have their timing constrained by a significant peptide, and all that are free variables temporally.  '0' is the value used to encode the inactive activity state.  There are very few inactive windows in the unconstrained proteins, but 12 of 143 are inactive at the first time point.

```python
constrainedProts = [row for row in protTiming if row['NodeType'] == 'SigPrize']
freeProts = [row for row in protTiming if (row['NodeType'] == 'InsigPrize' or row['NodeType'] == 'Steiner')]
allProts = list(constrainedProts)
allProts.extend(freeProts)
print '%d proteins in the Refiner pathway (%d constrained and %d free)' % (len(allProts), len(constrainedProts), len(freeProts))

allProtsTiming = dict()
freeProtsTiming = dict()
constrainedProtsTiming = dict()
for i in range(7):
    time = 2**(i+1)
    col = 'ActivitySummary%dMin' % time
    allProtsTiming[time] = sum([1 for row in allProts if not row[col] == '0'])
    freeProtsTiming[time] = sum([1 for row in freeProts if not row[col] == '0'])
    constrainedProtsTiming[time] = sum([1 for row in constrainedProts if not row[col] == '0'])
print 'All protein timing: %s' % str(allProtsTiming)
print 'Constrained protein timing: %s' % str(constrainedProtsTiming)
print 'Free protein timing: %s' % str(freeProtsTiming)
```

## Count the number of proteins that are first active at each time point

The number of 'Not active' proteins is not written to the file.  The 5 'Not active' proteins detected previously was an error, all proteins in the synthesized pathway are active at some time point.

```python
allProtsFirst = dict()
constrainedProtsFirst = dict()
for i in range(7):
    time = 2**(i+1)
    allProtsFirst[time] = sum([1 for row in allProts if row['FirstActive'] == str(time)])
    constrainedProtsFirst[time] = sum([1 for row in constrainedProts if row['FirstActive'] == str(time)])
print '%d proteins are not active' % sum([1 for row in allProts if row['FirstActive'] == "Not active"])
print '%d constrained proteins are not active' % sum([1 for row in constrainedProts if row['FirstActive'] == "Not active"])
```

## Write a text file showing the number and fraction of active peptides/proteins at each time point

```python
with open(timingTableFile, 'w') as timingF:
    timingF.write('Time\tActive total peptides\t')
    timingF.write('Active significant peptides\t')
    timingF.write('Active proteins\t')
    timingF.write('Active constrained proteins\t')
    timingF.write('Active free proteins\t')
    timingF.write('First active significant peptides\t')
    timingF.write('First active proteins\t')
    timingF.write('First active constrained proteins\n')
    
    for time in sorted(sigPeptidesTiming.keys()):
        timingF.write('%d min\t' % time)
        timingF.write('%d (%d%%)\t' % (sigPeptidesTiming[time], round(100*float(sigPeptidesTiming[time])/totPeptides)))
        timingF.write('%d (%d%%)\t' % (sigPeptidesTiming[time], round(100*float(sigPeptidesTiming[time])/sigPeptides)))
        timingF.write('%d (%d%%)\t' % (allProtsTiming[time], round(100*float(allProtsTiming[time])/len(allProts))))
        timingF.write('%d (%d%%)\t' % (constrainedProtsTiming[time], round(100*float(constrainedProtsTiming[time])/len(constrainedProts))))
        timingF.write('%d (%d%%)\t' % (freeProtsTiming[time], round(100*float(freeProtsTiming[time])/len(freeProts))))
        timingF.write('%d (%d%%)\t' % (sigPeptidesFirst[time], round(100*float(sigPeptidesFirst[time])/sigPeptides)))
        timingF.write('%d (%d%%)\t' % (allProtsFirst[time], round(100*float(allProtsFirst[time])/len(allProts))))
        timingF.write('%d (%d%%)\n' % (constrainedProtsFirst[time], round(100*float(constrainedProtsFirst[time])/len(constrainedProts))))
```
